```python
real_pm = pd.read_csv('/Users/tamsopanat/Desktop/MedCMU - FamMed/Dustboy/result/pm25.csv')
real_temp = pd.read_csv('../data2/temp.csv')
real_temp.loc[real_temp['amphoe'].str.contains('เมือง', na=False), 'amphoe'] = 'เมือง'

real_pm = real_pm[real_pm['province'].isin(real_temp['province'])]
real_pm = real_pm[real_pm['date'].isin(real_temp['date'])]
```

```python
real = real_pm.merge(real_temp, how='left', on=['datetime', 'date', 'hour', 'province', 'amphoe', 'tambon'])
mean_temps = real_temp.groupby(by=['datetime', 'date', 'hour'], as_index=False)['temperature'].mean()
mean_temps = mean_temps.rename(columns={'temperature': 'hourly_mean_temp'})
real = real.merge(mean_temps, how='left', on=['datetime', 'date', 'hour'])
real['temperature'] = real['temperature'].fillna(real['hourly_mean_temp'])
real = real.drop(columns=['hourly_mean_temp'])
real = real.rename(columns={'pm25': 'pm25_avg', 'temperature': 'temperature_avg'})
# real.to_csv('../data2/environment_hourly.csv', index=False)
real
```

```python
real_daily = real.drop(columns = ['source_id', 'datetime', 'hour', 'lat', 'lon']).groupby(by = ['date', 'province', 'amphoe', 'tambon'], as_index = False).mean()
real_daily.to_csv('../data2/environment_daily.csv', index = False)
real_daily
```

```python
# Define the columns to join on (adjust these if your daily data only uses 'date')
join_columns = ['date', 'province', 'amphoe', 'tambon']

# Merge area_alert_daily with real_daily
merged_daily = data['area_alert_daily'].merge(real_daily, how='left', on=join_columns)
merged_daily['avg_pm25'] = merged_daily['pm25_avg']
merged_daily['avg_temperature'] = merged_daily['temperature_avg']

merged_daily = merged_daily.drop(columns=['pm25_avg', 'temperature_avg'])

# Display the result
merged_daily = merged_daily.dropna()
merged_daily.to_csv('../data2/area_alert_daily.csv', index = False)
merged_daily
```
